```python
keys = [
    'Unit#0e31a2b3/leg2/leg2-hinge2',
    'Unit#a8ae2d23/leg2/leg2-hinge1',
    'Unit#a8ae2d23/'
]

named = np.empty((0,))
cached = np.empty((0,))


for key in keys:
    idx = sbenv.physics.named.data.qpos._convert_key(key)
    named = np.concatenate((named, sbenv.physics.named.data.qpos[key]))
    cached = np.concatenate((cached, sbenv.physics.data.qpos[idx]))

print(named)
print(cached)
print(named - cached)
```

```python
keys = [
    'Unit#0e31a2b3/leg2/leg2-hinge2',
    'Unit#a8ae2d23/leg2/leg2-hinge1',
    'Unit#a8ae2d23/'
]

indices = np.arange(len(sbenv.physics.data.qpos), dtype=int)

idx = np.empty((0,), dtype=int)
named = np.empty((0,))

for key in keys:
    idx_arr = indices[sbenv.physics.named.data.qpos.axes.row.convert_key_item(key)]
    idx = np.concatenate((idx, idx_arr))
    named = np.concatenate((named, sbenv.physics.named.data.qpos[key]))

cached = sbenv.physics.data.qpos[idx]

print(named)
print(cached)
print(named - cached)
```
